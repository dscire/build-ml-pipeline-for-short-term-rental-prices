Download and open the dataset from the corresponding W&B project.

```python
run = wandb.init(project="nyc_airbnb", group="eda", save_code=True)
local_path = wandb.use_artifact("sample.csv:latest").file()
df = pd.read_csv(local_path)
```

```python
# Take a look at the content of the dataset

df.head()
```

Import tool to analyze and provide a visualization of the data content.

NOTE: The original examples recommended to use "pandas_profiling" python package, but that seems to be obsolete and it has been replaced by "ydata_profiling" for this project.

```python
import ydata_profiling

profile = ydata_profiling.ProfileReport(df)
profile.to_widgets()
```

Perform a few cleanup steps, as recommended by the project's instructions.

```python
# Drop outliers
min_price = 10
max_price = 350
idx = df['price'].between(min_price, max_price)
df = df[idx].copy()
# Convert last_review to datetime
df['last_review'] = pd.to_datetime(df['last_review'])
```
